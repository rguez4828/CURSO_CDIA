## 1. Objetivos y alcances

El flujo de pacientes con afectaciones cardiacas en el Hospital anualmente es elevado y todos los datos son recogidos por expertos en el área clinica y cardiología. Como resultado de las consultas se realizan análisis de varios parámetros médicos que son necesarios para un diagnóstico final. Analizar los datos que son recogidos por todos los especialistas no es tarea fácil por lo que es necesario un mecanismo que ayude a la toma de decisiones.

**Objetivo General**

Predecir la probabilidad de que el paciente padezca una falla cardiaca que pueda llevar a su fallecimiento, esto se puede realizar mediante la calificación de la variable *D egreso*, por lo que el objetivo será:
* Predecir el diagnóstico del egreso del paciente.

````python
import missingno as msno

def analisis_valores_faltantes(df: pd.DataFrame, msno_bar = False):
    """
    Genera una tabla con el número y porcentaje de valores faltantes por columna en un DataFrame.

    Parámetros:
    df (pd. DataFrame): DataFrame de entrada.

    Retorna:
    pd. DataFrame: Tabla con columnas que tienen valores faltantes, indicando el número y porcentaje de valores faltantes.
    """
    faltantes_suma = df.isnull().sum() # total de valores faltantes
    faltantes_porc = 100 * (faltantes_suma / len(df)) # porcentaje de valores faltantes
    faltantes_df = pd.concat([faltantes_suma, faltantes_porc], axis=1) # tabla de resultados
    faltantes_df = faltantes_df.rename( # renombramos columnas
        columns = {0 : 'Valores faltantes', 1 : '% del total de valores'}, )

    faltantes_df = faltantes_df[ # ordenar
            faltantes_df.iloc[:,1] != 0].sort_values(
        '% del total de valores', ascending=False).round(1)
    print ("El Dataframe tiene " + str(df.shape[1]) + " columnas.\n"
            "Hay " + str(faltantes_df.shape[0]) +
              " Columnas que tienen valores faltantes.")

    if msno_bar:
        msno.bar(df)

    return faltantes_df
````

A continuación se detallan 4 funciones que se crearon para facilitar el proceso de análisis univariado y bivariado:
1. **analisis_var_cuantitativa_sns:**   Analiza una variable cuantitativa específica en el dataframe calculando medidas estadísticas como la media,
                                        moda y mediana. Determina la asimetría de la distribución e imprime un mensaje indicando si la distribución
                                        es simétrica, asimétrica positiva o asimétrica negativa. Además, visualiza la distribución utilizando un histograma
                                        con una línea de densidad y un diagrama de cajas.
2. **analisis_var_cualitativa_sns:**    Analiza una variable cualitativa específica en el dataframe creando un gráfico de conteo utilizando Seaborn.
                                        Si se proporciona un diccionario de mapeo, utiliza este diccionario para codificar los valores de la variable antes
                                        de generar el gráfico.
3. **analisis_atipicos:**               Analiza una variable específica en el dataframe para identificar valores atípicos utilizando los métodos IQR (Rango Intercuartil)
                                        y Z-score. Si se especifica, elimina los valores atípicos relevantes.
4. **normalizar_var_cuantitativa:**     Crea una nueva columna en el dataframe normalizando la variable _x utilizando RobustScaler si hay valores atípicos,
                                        o MinMaxScaler si no hay valores atípicos. La nueva columna se nombra como ROBUST_x o MINMAX_x según corresponda.

Realizaremos el análisis univariado y bivariado de las variables cuantitativas y cualitativas.

````python
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
import numpy as np

sns.set_theme()
sns.set_style()

def analisis_var_cuantitativa_sns(dataframe: pd.DataFrame, _x: str):
    """
    Analiza una variable cuantitativa específica en el dataframe calculando medidas estadísticas como la media,
    moda y mediana. Determina la asimetría de la distribución e imprime un mensaje indicando si la distribución
    es simétrica, asimétrica positiva o asimétrica negativa. Además, visualiza la distribución utilizando un histograma
    con una línea de densidad y un diagrama de cajas.

    :param dataframe: El DataFrame de pandas que contiene los datos a analizar.
    :param _x: Una cadena representando el nombre de la columna en el dataframe que se va a analizar.
    :return: None
    """
    mean_df = dataframe[_x].mean()
    mode_df = dataframe[_x].mode()[0]
    median_df = dataframe[_x].median()

    if mean_df > median_df and mean_df > mode_df:
        display(f"La distribución de la variable {_x} tiene asimetría positiva (cola hacia la derecha)")  # Media > Mediana > Moda
    elif mean_df < median_df and mean_df < mode_df:
        display(f"La distribución de la variable {_x} tiene asimetría negativa (cola hacia la izquierda)")  # Moda > Mediana > Media
    else:
        display(f"La distribución de la variable {_x} parece ser simétrica (Media ≈ Mediana ≈ Moda)")  # Asimetría cero

    _fig, _axs = plt.subplots(1, 2, figsize=(15, 3))

    # Creating a countplot to visualize the distribution of 'Outcome' values
    sns.histplot(x=_x, data=dataframe, kde=True, ax=_axs[0])
    sns.boxplot(x=_x, data=dataframe, ax=_axs[1])

    plt.show()

def analisis_var_cualitativa_sns(dataframe: pd.DataFrame, _x: str, _map: dict = None, rot = 0):
    """
    Analiza una variable cualitativa específica en el dataframe creando un gráfico de conteo utilizando Seaborn.
    Si se proporciona un diccionario de mapeo, utiliza este diccionario para codificar los valores de la variable antes
    de generar el gráfico.

    :param dataframe: El DataFrame de pandas que contiene los datos a analizar.
    :param _x: Una cadena representando el nombre de la columna en el dataframe que se va a analizar.
    :param _map: Un diccionario opcional utilizado para mapear valores de la variable especificada para una mejor
                 visualización o codificación. Si no se proporciona, se utiliza directamente la variable original.
    :param rot: Rotar etiquetas del eje X.
    :return: None
    """
    _fig, _axs = plt.subplots(figsize=(15, 3))
    _axs.set_xlabel(_x)

    if rot:
        _axs.tick_params(axis='x', rotation=rot)

    if _map is None:
        sns.countplot(data=dataframe, x=_x, ax=_axs)
    else:
        dataframe.loc[:, f'{_x}_label'] = dataframe[_x].map(_map)
        sns.countplot(data=dataframe, x=f'{_x}_label', ax=_axs, )

def analisis_atipicos(dataframe: pd.DataFrame, _x: str, del_outliers: bool):
    """
    Analiza una variable específica en el dataframe para identificar valores atípicos utilizando los métodos IQR (Rango Intercuartil)
    y Z-score. Si se especifica, elimina los valores atípicos relevantes.

    :param dataframe: El DataFrame de pandas que contiene los datos a analizar.
    :param _x: Una cadena representando el nombre de la columna en el dataframe que se va a analizar para detectar outliers.
    :param del_outliers: Un booleano que indica si se deben eliminar los valores atípicos identificados del dataframe.
    :return: None
    """
    mean_df = dataframe[_x].mean()
    median_df = dataframe[_x].median()
    std_df = dataframe[_x].std()

    dataframe[_x].sort_values()
    _Q1 = dataframe[_x].quantile(0.25)
    _Q3 = dataframe[_x].quantile(0.75)

    _IQR = _Q3 - _Q1 # Rango intercuartil

    _limite_inferior = _Q1 - 1.5 * _IQR
    _limite_superior = _Q3 + 1.5 * _IQR

    print(f'--------- Variable {_x} ---------')
    print(f'La media es: {mean_df}, la mediana es: {median_df} y La desviación estandar es: {std_df}')

    print(f'El Cuartil Q1 es: {_Q1}, el Cuartil Q3 es: {_Q3} y el Rango intercuartil es: {_IQR}, con un Limite inferior de: {_limite_inferior} y un Limite superior de {_limite_superior}')

    _escalado_puntoZ = StandardScaler()
    dataframe[f'Z_{_x}'] = _escalado_puntoZ.fit_transform(dataframe[[_x]])

    print('Media {} y desvío estándar {} utilizados '. format(_escalado_puntoZ.mean_[0], np.sqrt(_escalado_puntoZ.var_)[0]))

    z_umbral = 3

    atipicos_dataframe = dataframe.loc[(dataframe[f'Z_{_x}']< -(z_umbral)) | (dataframe[f'Z_{_x}']> z_umbral),]

    display("Atípicos por puntaje Z :",atipicos_dataframe[[_x,f'Z_{_x}']].sort_values(by=f'Z_{_x}'))

    if del_outliers:
        print('Se eliminarán los valores atípicos relevantes después del análisis con puntaje Z.')
        print(f'Solo se eliminarán los que se muestran en la tabla anterior, no los que están fuera del Limite inferior de: {_limite_inferior} y el Limite superior de {_limite_superior}')
        # dataframe = dataframe[(dataframe[_x] >= _limite_inferior) & (dataframe[_x]  <= _limite_superior)] # Esta linea elimina los valores atípicos fuera del IQR
        dataframe.drop(atipicos_dataframe.index, inplace=True)

def normalizar_var_cuantitativa(dataframe: pd.DataFrame, _x: str, outliers: bool):
    """
    Crea una nueva columna en el dataframe normalizando la variable _x utilizando RobustScaler si hay valores atípicos,
    o MinMaxScaler si no hay valores atípicos. La nueva columna se nombra como ROBUST_x o MINMAX_x según corresponda.

    :param dataframe: El DataFrame de pandas que contiene los datos a normalizar.
    :param _x: Una cadena representando el nombre de la columna en el dataframe que se va a normalizar.
    :param outliers: Un booleano indicando si la variable contiene valores atípicos.
    :return: None
    """
    if outliers:
        escalado_Robust = RobustScaler()
        dataframe[f'ROBUST_{_x}'] = escalado_Robust.fit_transform(dataframe[[_x]])
    else:
        escalado_minMax = MinMaxScaler() # Inicializar la clase MinMaxScaler
        dataframe[f'MINMAX_{_x}'] = escalado_minMax.fit_transform(dataframe[[_x]])

def analisis_var_objetivo(_dataframe: pd.DataFrame, _etiqueta: str, _caracteristica: list, _counterplot = True, _map: dict[int, str] = None):
    """
    Realiza un análisis completo de la variable objetivo y sus relaciones con otras variables.

    Esta función proporciona un análisis descriptivo de la variable objetivo, incluyendo:
    - Distribución de frecuencias relativas
    - Visualización mediante gráficos de barras
    - Estadísticas descriptivas agrupadas por la variable objetivo
    - Matriz de gráficos de dispersión (pairplot) para analizar relaciones

    Parameters:
    -----------
    _dataframe: pd.DataFrame
        DataFrame que contiene los datos a analizar
    _etiqueta: str
        Nombre de la columna que representa la variable objetivo
    _caracteristica: list
        Lista de nombres de columnas de características para incluir en el pairplot
    _counterplot: bool, optional (default=True)
        Si es True, muestra el análisis descriptivo completo de la variable objetivo
    _map: dict[int, str], optional (default=None)
        Diccionario para mapear valores numéricos a etiquetas textuales en las visualizaciones

    Returns:
    --------
    None
        La función muestra resultados en consola y genera visualizaciones, pero no retorna valores

    Notas:
    ------
    - La función modifica temporalmente las opciones de pandas para mostrar todas las columnas y filas
    - Añade automáticamente la variable objetivo a la lista de características para el pairplot
    - Requiere que existan las funciones 'analisis_var_cualitativa_sns' y las librerías pandas, seaborn y matplotlib
    """
    if _counterplot:
        print(_dataframe[_etiqueta].value_counts("%")*100)
        analisis_var_cualitativa_sns(_dataframe, _etiqueta, _map)

        pd.set_option('display.max_columns', None)
        pd.set_option('display.max_rows', None)

        # Agrupar los datos por la columna 'Resultado' y calcular la media para cada grupo
        display(_dataframe.groupby(_etiqueta).describe())

    _c = _caracteristica.copy()
    _c.append(_etiqueta)

    # Generar colores aleatorios según la cantidad de etiquetas únicas
    unique_labels = _dataframe[_etiqueta].nunique()
    import random

    # Generar colores aleatorios
    colors = []
    for _ in range(unique_labels):
        # Generar un color RGB aleatorio
        color = "#{:06x}".format(random.randint(0, 0xFFFFFF))
        colors.append(color)

    sns.pairplot(_dataframe[_c], hue = _etiqueta, diag_kind="hist", palette=colors)
    # sns.pairplot(_dataframe[_c], hue = _etiqueta, diag_kind="hist")

    plt.show()
````

Para en análisis de las variables cualitativas se analizará la correlación de la variable objetivo con las variables cualitativas relevantes. Para esto se crearon dos funciones scatter_hist() y plot_multiple_scatter_hist().

````python
def scatter_hist(x, y, ax, ax_histx, ax_histy):
    """
    Crea un gráfico de dispersión con histogramas marginales.

    Parameters:
    -----------
    x: pd.DataFrame
        DataFrame con una columna para el eje x
    y: pd.DataFrame
        DataFrame con una columna para el eje y
    ax: matplotlib.axes.Axes
        Ejes para el gráfico de dispersión
    ax_histx: matplotlib.axes.Axes
        Ejes para el histograma superior
    ax_histy: matplotlib.axes.Axes
        Ejes para el histograma derecho
    """
    # no labels
    ax_histx.tick_params(axis="x", labelbottom=False)
    ax_histy.tick_params(axis="y", labelleft=False)

    x_label = x.columns[0]  # Tomamos el nombre de la columna como etiqueta
    y_label = y.columns[0]

    # Determinar límites comunes para x e y
    x_min, x_max = x.min().values[0], x.max().values[0]
    y_min, y_max = y.min().values[0], y.max().values[0]

    common_min = min(x_min, y_min) - 10
    common_max = max(x_max, y_max) + 10

    # Gráfico de dispersión
    ax.scatter(x, y, c='skyblue', alpha=0.6)
    ax.set_xlabel(x_label)
    ax.set_ylabel(y_label)

    # Establecer límites iguales para x e y en el scatter
    ax.set_xlim(common_min, common_max)
    ax.set_ylim(common_min, common_max)

    # now determine nice limits by hand:
    binwidth = 0.25
    xymax = max(np.max(np.abs(x)), np.max(np.abs(y)))
    lim = (int(xymax/binwidth) + 1) * binwidth

    bins = np.arange(-lim, lim + binwidth, binwidth)
    ax_histx.hist(x, bins=bins, edgecolor='salmon', color='lightgray', alpha=0.6)
    ax_histx.set_xlim(common_min, common_max)  # Ajustar límite del histograma X
    ax_histy.hist(y, bins=bins, orientation='horizontal', edgecolor='salmon', color='lightgray', alpha=0.6)
    ax_histy.set_ylim(common_min, common_max)  # Ajustar límite del histograma Y

def plot_multiple_scatter_hist(x_list, y, dataframe):
    """
    Crea múltiples gráficos de dispersión con histogramas marginales para una lista de variables x
    contra una variable y utilizando subplot_mosaic en un layout de dos columnas.

    Parameters:
    -----------
    x_list : list
        Lista de nombres de columnas para usar como variables x
    y : str
        Nombre de la columna para usar como variable y
    dataframe : pd.DataFrame
        DataFrame que contiene los datos
    """
    num_plots = len(x_list)

    # Calcular número de filas necesarias para dos columnas
    num_rows = (num_plots + 1) // 2  # Redondeo hacia arriba

    # Crear el layout usando subplot_mosaic con dos columnas
    mosaic_layout = []
    axes_mapping = {}

    # Construir el layout de dos columnas
    plot_index = 0
    for row in range(num_rows):
        # Primera columna
        if plot_index < num_plots:
            histx_name_0 = f'histx_{row}_0'
            empty_name_0 = f'.'
            scatter_name_0 = f'scatter_{row}_0'
            histy_name_0 = f'histy_{row}_0'

            # Segunda columna (si existe)
            if plot_index + 1 < num_plots:
                histx_name_1 = f'histx_{row}_1'
                empty_name_1 = f'.'
                scatter_name_1 = f'scatter_{row}_1'
                histy_name_1 = f'histy_{row}_1'

                # Añadir filas con dos columnas
                mosaic_layout.append([histx_name_0, empty_name_0, '.', histx_name_1, empty_name_1])
                mosaic_layout.append([scatter_name_0, histy_name_0, '.', scatter_name_1, histy_name_1])
            else:
                # Última fila con solo una columna
                mosaic_layout.append([histx_name_0, empty_name_0, '.', '.', '.'])
                mosaic_layout.append([scatter_name_0, histy_name_0, '.', '.', '.'])

            # Mapear los índices a los nombres de los ejes
            axes_mapping[plot_index] = {
                'histx': histx_name_0,
                'scatter': scatter_name_0,
                'histy': histy_name_0
            }

            if plot_index + 1 < num_plots:
                axes_mapping[plot_index + 1] = {
                    'histx': histx_name_1,
                    'scatter': scatter_name_1,
                    'histy': histy_name_1
                }

            plot_index += 2

        # Añadir fila vacía como separador entre filas si no es la última fila
        if row < num_rows - 1:
            mosaic_layout.append(['.', '.', '.', '.', '.'])

    # Calcular el tamaño de la figura
    fig_width = 16  # Ancho mayor para acomodar dos columnas
    fig_height = max(8, 5 * num_rows)

    # Crear la figura con el mosaic
    fig, axs = plt.subplot_mosaic(mosaic_layout,
                                  figsize=(fig_width, fig_height),
                                  width_ratios=[4, 1, 0.1, 4, 1],  # Proporciones: [plot, hist, separador, plot, hist]
                                  layout='constrained')

    # Crear un gráfico para cada variable x
    for i, x_col in enumerate(x_list):
        # Seleccionar los datos
        x_data = dataframe[[x_col]]
        y_data = dataframe[[y]]

        # Obtener los nombres de los ejes para este plot
        axis_names = axes_mapping[i]

        # Crear el gráfico
        scatter_hist(x_data, y_data,
                    axs[axis_names['scatter']],
                    axs[axis_names['histx']],
                    axs[axis_names['histy']])

    plt.show()
````

````python
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score, f1_score
from sklearn. metrics import classification_report

def _coronario_confusion_matrix():
    """
    Genera y visualiza la matriz de confusión para el clasificador de árbol de decisión.

    Esta función calcula la matriz de confusión comparando las predicciones del modelo
    de árbol de decisión con los valores reales del conjunto de prueba, y muestra
    una representación gráfica heatmap de la matriz con anotaciones.

    Global Variables:
    -----------------
    _y_pred_arbolDec : array-like
        Predicciones del modelo de árbol de decisión - debe estar definido globalmente
    _y_test : array-like
        Valores reales/etiquetas verdaderas del conjunto de prueba - debe estar definido globalmente
    letiqueta_map : list
        Lista con los nombres de las clases mapeadas para las etiquetas de los ejes

    Returns:
    --------
    None
        La función no retorna valores explícitamente, pero muestra una visualización

    Visualizations:
    ---------------
    - Heatmap de la matriz de confusión con:
      * Valores anotados en cada celda
      * Colormap 'Blues' para mejor visualización
      * Etiquetas de clase en ambos ejes
      * Título descriptivo
      * Etiquetas de ejes X e Y

    Matriz de Confusión:
    --------------------
    La matriz muestra:
    - Eje Y (filas): Valores reales/actuales
    - Eje X (columnas): Valores predichos
    - Cada celda: Número de instancias clasificadas en esa combinación

    Example:
    --------
    >>> # Variables globales deben estar definidas previamente
    >>> _coronario_confusion_matrix()
    """
    global _y_pred_arbolDec
    # Matriz de confusion KNN
    _cm_arbolDec = confusion_matrix(_y_test, _y_pred_arbolDec)

    fig, ax = plt.subplots(1, 1, figsize=(15, 5))

    sns.heatmap(_cm_arbolDec, annot=True, fmt='d', cmap='Blues', ax=ax, xticklabels=letiqueta_map, yticklabels=letiqueta_map)
    # sns.heatmap(_cm_arbolDec, annot=True, fmt='d', cmap='Blues', ax=ax)
    ax.set_title("Matriz de confusión de árbol de decisión")
    ax.set_xlabel("Valores Predichos")
    ax.set_ylabel("Valores Actuales o Reales")

    plt.show()
    # plt.tight_layout()

def _coronario_metricas(_ac: True, _pr: False, _re : False, _f1: False, _average: str = None):
    """
    Calcula y muestra las métricas de evaluación del modelo de árbol de decisión.

    Esta función permite calcular y mostrar métricas específicas de rendimiento del modelo
    según los parámetros booleanos proporcionados. Las métricas incluyen Accuracy, Precision,
    Recall y F1-Score, tanto por clase como promediadas.

    Parameters:
    -----------
    _ac : bool, optional (default=True)
        Si es True, muestra la métrica de Accuracy promedio
    _pr : bool, optional (default=False)
        Si es True, muestra la métrica de Precision por clase y opcionalmente promediada
    _re : bool, optional (default=False)
        Si es True, muestra la métrica de Recall por clase y opcionalmente promediada
    _f1 : bool, optional (default=False)
        Si es True, muestra la métrica F1-Score por clase y opcionalmente promediada
    _average : str, optional (default=None)
        Tipo de promedio a calcular para Precision, Recall y F1-Score.
        Valores posibles: 'micro', 'macro', 'weighted', 'samples' o None.
        Si es None, solo se muestran las métricas por clase.

    Global Variables:
    -----------------
    _y_pred_arbolDec : array-like
        Predicciones del modelo de árbol de decisión - debe estar definido globalmente
    _y_test : array-like
        Valores reales/etiquetas verdaderas del conjunto de prueba - debe estar definido globalmente
    letiqueta_map : list
        Lista con los nombres de las clases para el classification report

    Returns:
    --------
    None
        La función no retorna valores, solo imprime las métricas en consola

    Prints:
    -------
    - Accuracy promedio (si _ac=True)
    - Precision por clase y promedio (si _pr=True)
    - Recall por clase y promedio (si _re=True)
    - F1-Score por clase y promedio (si _f1=True)
    - Classification report completo con todas las métricas por clase

    Example:
    --------
    >>> # Mostrar solo accuracy y classification report
    >>> _coronario_metricas(True, False, False, False)
    >>>
    >>> # Mostrar todas las métricas con promedio ponderado
    >>> _coronario_metricas(True, True, True, True, 'weighted')
    >>>
    >>> # Mostrar métricas por clase sin promedio
    >>> _coronario_metricas(True, True, True, True)
    """
    global _y_pred_arbolDec
    if _ac:
        print('Accuracy promedio: ', accuracy_score(_y_test, _y_pred_arbolDec))

    if _pr:
        print('En general Precision: ', precision_score(_y_test, _y_pred_arbolDec, average=None))
        if _average:
            print('Precision promedio: ', precision_score(_y_test, _y_pred_arbolDec, average=_average))

    if _re:
        print('En general Recall: ', recall_score(_y_test, _y_pred_arbolDec, average=None))
        if _average:
            print('Recall promedio: ', recall_score(_y_test, _y_pred_arbolDec, average=_average))

    if _f1:
        print('En general F-Score (F1): ', f1_score(_y_test, _y_pred_arbolDec, average=None))
        if _average:
            print('F-Score (F1) promedio: ', f1_score(_y_test, _y_pred_arbolDec, average=_average))

    print(classification_report(_y_test, _y_pred_arbolDec, target_names=letiqueta_map))
````
